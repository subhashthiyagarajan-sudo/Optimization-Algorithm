```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from keras.models import Sequential
from keras.layers import Dense, GRU
from keras.callbacks import EarlyStopping
import tensorflow as tf

from google.colab import drive
drive.mount('/content/drive')

# Load and preprocess data
file_path = "/content/drive/MyDrive/data/TECHM.NS.csv"
data = pd.read_csv(file_path)

data['Date'] = pd.to_datetime(data['Date'])
data.set_index('Date', inplace=True)

features = data[['Open', 'High', 'Low', 'Close']]
target = data['Close']

scaler = MinMaxScaler(feature_range=(0, 1))
scaled_features = scaler.fit_transform(features.astype(np.float32))
scaled_target = scaler.fit_transform(target.values.reshape(-1, 1).astype(np.float32))

train_size = int(len(data) * 0.8)
train_features, test_features = scaled_features[:train_size], scaled_features[train_size:]
train_target, test_target = scaled_target[:train_size], scaled_target[train_size:]

# Reshape for GRU [samples, timesteps, features]
train_features = train_features.reshape((train_features.shape[0], 1, train_features.shape[1]))
test_features = test_features.reshape((test_features.shape[0], 1, test_features.shape[1]))

def fuzzy_rough_set(features):
    # Dummy function; replace with your own fuzzy rough set logic if needed
    return features

train_features = fuzzy_rough_set(train_features)
test_features = fuzzy_rough_set(test_features)

# Build and evaluate GRU model based on provided hyperparameters
def build_and_evaluate_model(params):
    units1, units2, batch_size = map(int, params)
    model = Sequential([
        GRU(units1, return_sequences=True, input_shape=(train_features.shape[1], train_features.shape[2]), dtype=tf.float32),
        GRU(units2, dtype=tf.float32),
        Dense(1, dtype=tf.float32)
    ])
    model.compile(optimizer='adam', loss='mean_squared_error')
    early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    model.fit(train_features, train_target, epochs=50, batch_size=batch_size,
              validation_data=(test_features, test_target), callbacks=[early_stopping], verbose=0)
    predictions = scaler.inverse_transform(model.predict(test_features))
    actuals = scaler.inverse_transform(test_target)
    mse = mean_squared_error(actuals, predictions)
    return mse  # GWO will try to minimize this value

# Grey Wolf Optimization (GWO) function
def gwo(obj_function, lb, ub, n_wolves=10, max_iter=5):
    dim = len(lb)
    # Initialize wolf positions
    wolves = np.random.uniform(lb, ub, (n_wolves, dim)).astype(np.float32)
    # Evaluate fitness of each wolf
    fitness = np.array([obj_function(wolf) for wolf in wolves])
    # Identify the best three wolves: alpha, beta, delta
    sorted_indices = np.argsort(fitness)
    alpha = wolves[sorted_indices[0]].copy()
    beta = wolves[sorted_indices[1]].copy()
    delta = wolves[sorted_indices[2]].copy()

    for it in range(max_iter):
        a = 2 - 2 * (it / max_iter)  # linearly decreasing parameter a from 2 to 0
        for i in range(n_wolves):
            # Calculate the coefficients for alpha, beta, and delta
            r1 = np.random.rand(dim)
            r2 = np.random.rand(dim)
            A1 = 2 * a * r1 - a
            C1 = 2 * r2
            D_alpha = abs(C1 * alpha - wolves[i])
            X1 = alpha - A1 * D_alpha

            r1 = np.random.rand(dim)
            r2 = np.random.rand(dim)
            A2 = 2 * a * r1 - a
            C2 = 2 * r2
            D_beta = abs(C2 * beta - wolves[i])
            X2 = beta - A2 * D_beta

            r1 = np.random.rand(dim)
            r2 = np.random.rand(dim)
            A3 = 2 * a * r1 - a
            C3 = 2 * r2
            D_delta = abs(C3 * delta - wolves[i])
            X3 = delta - A3 * D_delta

            # Update wolf position as the average of the three updated positions
            wolves[i] = (X1 + X2 + X3) / 3
            # Clip the wolf position within the bounds
            wolves[i] = np.clip(wolves[i], lb, ub)

        # Update fitness values
        fitness = np.array([obj_function(wolf) for wolf in wolves])
        sorted_indices = np.argsort(fitness)
        alpha = wolves[sorted_indices[0]].copy()
        beta = wolves[sorted_indices[1]].copy()
        delta = wolves[sorted_indices[2]].copy()
        best_fitness = fitness[sorted_indices[0]]
        print(f"Iteration {it+1}/{max_iter}, Best MSE: {best_fitness}")

    return alpha, best_fitness

# Define lower and upper bounds for hyperparameters: [units1, units2, batch_size]
lb = np.array([10, 10, 16], dtype=np.float32)
ub = np.array([100, 100, 128], dtype=np.float32)

optimal_params, _ = gwo(build_and_evaluate_model, lb, ub, n_wolves=10, max_iter=5)
units1_optimal, units2_optimal, batch_size_optimal = map(int, optimal_params)

print("Optimal Parameters:")
print("Units1:", units1_optimal)
print("Units2:", units2_optimal)
print("Batch Size:", batch_size_optimal)

# Build final GRU model using optimal hyperparameters
model = Sequential([
    GRU(units1_optimal, return_sequences=True, input_shape=(train_features.shape[1], train_features.shape[2]), dtype=tf.float32),
    GRU(units2_optimal, dtype=tf.float32),
    Dense(1, dtype=tf.float32)
])
model.compile(optimizer='adam', loss='mean_squared_error')

early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
history = model.fit(train_features, train_target, epochs=50, batch_size=batch_size_optimal,
                    validation_data=(test_features, test_target), callbacks=[early_stopping], verbose=1)

# Predictions and performance evaluation
predictions = scaler.inverse_transform(model.predict(test_features))
actuals = scaler.inverse_transform(test_target)

mae = mean_absolute_error(actuals, predictions)
rmse = np.sqrt(mean_squared_error(actuals, predictions))
r2 = r2_score(actuals, predictions)

print(f"MAE: {mae}")
print(f"RMSE: {rmse}")
print(f"R^2: {r2}")

# Plot training and validation loss
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()

# Plot actual vs predicted prices
plt.figure(figsize=(10, 6))
plt.plot(actuals, label='Actual Prices')
plt.plot(predictions, label='Predicted Prices')
plt.title('Actual vs Predicted Prices')
plt.xlabel('Time Steps')
plt.ylabel('Price')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from keras.models import Sequential
from keras.layers import Dense, GRU
from keras.callbacks import EarlyStopping
import tensorflow as tf

from google.colab import drive
drive.mount('/content/drive')

# Load and preprocess data
file_path = "/content/drive/MyDrive/data/TECHM.NS.csv"
data = pd.read_csv(file_path)

data['Date'] = pd.to_datetime(data['Date'])
data.set_index('Date', inplace=True)

features = data[['Open', 'High', 'Low', 'Close']]
target = data['Close']

scaler = MinMaxScaler(feature_range=(0, 1))
scaled_features = scaler.fit_transform(features.astype(np.float32))
scaled_target = scaler.fit_transform(target.values.reshape(-1, 1).astype(np.float32))

train_size = int(len(data) * 0.8)
train_features, test_features = scaled_features[:train_size], scaled_features[train_size:]
train_target, test_target = scaled_target[:train_size], scaled_target[train_size:]

# Reshape for GRU [samples, timesteps, features]
train_features = train_features.reshape((train_features.shape[0], 1, train_features.shape[1]))
test_features = test_features.reshape((test_features.shape[0], 1, test_features.shape[1]))

def fuzzy_rough_set(features):
    # Dummy function; replace with your own fuzzy rough set logic if needed
    return features

train_features = fuzzy_rough_set(train_features)
test_features = fuzzy_rough_set(test_features)

# Build and evaluate GRU model based on provided hyperparameters
def build_and_evaluate_model(params):
    units1, units2, batch_size = map(int, params)
    model = Sequential([
        GRU(units1, return_sequences=True, input_shape=(train_features.shape[1], train_features.shape[2]), dtype=tf.float32),
        GRU(units2, dtype=tf.float32),
        Dense(1, dtype=tf.float32)
    ])
    model.compile(optimizer='adam', loss='mean_squared_error')
    early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    model.fit(train_features, train_target, epochs=50, batch_size=batch_size,
              validation_data=(test_features, test_target), callbacks=[early_stopping], verbose=0)
    predictions = scaler.inverse_transform(model.predict(test_features))
    actuals = scaler.inverse_transform(test_target)
    mse = mean_squared_error(actuals, predictions)
    return mse  # The Cuckoo Search will try to minimize this value

# Cuckoo Search Optimization function
def cuckoo_search(obj_function, lb, ub, n_nests=10, max_iter=5, pa=0.25):
    dim = len(lb)
    # Initialize nests (candidate solutions)
    nests = np.random.uniform(lb, ub, (n_nests, dim)).astype(np.float32)
    fitness = np.array([obj_function(nest) for nest in nests])

    # Identify the best nest
    best_index = np.argmin(fitness)
    best_nest = nests[best_index].copy()

    for it in range(max_iter):
        # Generate new solutions via random walk (approximate Levy flights)
        for i in range(n_nests):
            step_size = 0.01 * (ub - lb)  # step size can be tuned further
            new_solution = nests[i] + np.random.uniform(-1, 1, dim) * step_size
            new_solution = np.clip(new_solution, lb, ub)
            new_fitness = obj_function(new_solution)
            # Replace a random nest if the new solution is better
            j = np.random.randint(n_nests)
            if new_fitness < fitness[j]:
                nests[j] = new_solution
                fitness[j] = new_fitness

        # Abandon a fraction (pa) of the worst nests and reinitialize them
        num_abandon = int(pa * n_nests)
        worst_indices = np.argsort(fitness)[-num_abandon:]
        nests[worst_indices] = np.random.uniform(lb, ub, (num_abandon, dim)).astype(np.float32)
        for idx in worst_indices:
            fitness[idx] = obj_function(nests[idx])

        best_index = np.argmin(fitness)
        best_nest = nests[best_index].copy()
        print(f"Iteration {it+1}/{max_iter}, Best MSE: {fitness[best_index]}")

    return best_nest, fitness[best_index]

# Define lower and upper bounds for hyperparameters: [units1, units2, batch_size]
lb = np.array([10, 10, 16], dtype=np.float32)
ub = np.array([100, 100, 128], dtype=np.float32)

optimal_params, _ = cuckoo_search(build_and_evaluate_model, lb, ub, n_nests=10, max_iter=5, pa=0.25)
units1_optimal, units2_optimal, batch_size_optimal = map(int, optimal_params)

print("Optimal Parameters:")
print("Units1:", units1_optimal)
print("Units2:", units2_optimal)
print("Batch Size:", batch_size_optimal)

# Build final GRU model using optimal hyperparameters
model = Sequential([
    GRU(units1_optimal, return_sequences=True, input_shape=(train_features.shape[1], train_features.shape[2]), dtype=tf.float32),
    GRU(units2_optimal, dtype=tf.float32),
    Dense(1, dtype=tf.float32)
])
model.compile(optimizer='adam', loss='mean_squared_error')

early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
history = model.fit(train_features, train_target, epochs=50, batch_size=batch_size_optimal,
                    validation_data=(test_features, test_target), callbacks=[early_stopping], verbose=1)

# Predictions and performance evaluation
predictions = scaler.inverse_transform(model.predict(test_features))
actuals = scaler.inverse_transform(test_target)

mae = mean_absolute_error(actuals, predictions)
rmse = np.sqrt(mean_squared_error(actuals, predictions))
r2 = r2_score(actuals, predictions)

print(f"MAE: {mae}")
print(f"RMSE: {rmse}")
print(f"R^2: {r2}")

# Plot training and validation loss
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()

# Plot actual vs predicted prices
plt.figure(figsize=(10, 6))
plt.plot(actuals, label='Actual Prices')
plt.plot(predictions, label='Predicted Prices')
plt.title('Actual vs Predicted Prices')
plt.xlabel('Time Steps')
plt.ylabel('Price')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from keras.models import Sequential
from keras.layers import Dense, GRU
from keras.callbacks import EarlyStopping
import random

# Mount Google Drive (if using Google Colab)
from google.colab import drive
drive.mount('/content/drive')

# Load the dataset - Verify this path is correct after mounting your Drive!
file_path = "/content/drive/MyDrive/data/TECHM.NS.csv"
data = pd.read_csv(file_path)

# Preprocess the data
data['Date'] = pd.to_datetime(data['Date'])
data.set_index('Date', inplace=True)

# Select the features and target
features = data[['Open', 'High', 'Low', 'Close']]
target = data['Close']

# Scale the features
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_features = scaler.fit_transform(features)
scaled_target = scaler.fit_transform(target.values.reshape(-1, 1))

# Split the data into training and testing sets
train_size = int(len(data) * 0.8)
train_features, test_features = scaled_features[:train_size], scaled_features[train_size:]
train_target, test_target = scaled_target[:train_size], scaled_target[train_size:]

# Reshape data to fit GRU input requirements [samples, time steps, features]
train_features = train_features.reshape((train_features.shape[0], 1, train_features.shape[1]))
test_features = test_features.reshape((test_features.shape[0], 1, test_features.shape[1]))

# Placeholder function for fuzzy rough sets
def fuzzy_rough_set(features):
    # Implement your fuzzy rough set logic here
    return features

# Apply fuzzy rough set on features
train_features = fuzzy_rough_set(train_features)
test_features = fuzzy_rough_set(test_features)

# Define the objective function for Cassowaries Optimization Algorithm
def build_and_evaluate_model(params):
    units1 = int(params[0])
    units2 = int(params[1])
    batch_size = int(params[2])

    # Build the GRU model
    model = Sequential()
    model.add(GRU(units=units1, return_sequences=True, input_shape=(train_features.shape[1], train_features.shape[2])))
    model.add(GRU(units=units2))
    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mean_squared_error')

    # Train the model
    early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    history = model.fit(
        train_features,
        train_target,
        epochs=50,
        batch_size=batch_size,
        validation_data=(test_features, test_target),
        callbacks=[early_stopping],
        verbose=0
    )

    # Make predictions
    predictions = model.predict(test_features)
    actuals = scaler.inverse_transform(test_target)
    predictions = scaler.inverse_transform(predictions)

    # Calculate metrics
    mse = mean_squared_error(actuals, predictions)

    return mse  # COA will minimize this value

# Cassowaries Optimization Algorithm (COA) Implementation
def cassowaries_optimization(obj_function, lb, ub, n_agents=10, max_iter=5):
    agents = [np.random.uniform(lb, ub) for _ in range(n_agents)]
    fitness = [obj_function(agent) for agent in agents]

    for iteration in range(max_iter):
        for i in range(n_agents):
            step = np.random.uniform(-1, 1, size=len(lb)) * (ub - lb) * (1 - iteration / max_iter)
            new_agent = agents[i] + step
            new_agent = np.clip(new_agent, lb, ub)

            new_fitness = obj_function(new_agent)
            if new_fitness < fitness[i]:
                agents[i] = new_agent
                fitness[i] = new_fitness

    best_agent = agents[np.argmin(fitness)]
    best_fitness = min(fitness)
    return best_agent, best_fitness

# Define bounds for hyperparameters
lb = [10, 10, 16]  # Lower bounds for units1, units2, batch_size
ub = [100, 100, 128]  # Upper bounds for units1, units2, batch_size

# Run Cassowaries Optimization Algorithm to find optimal parameters
optimal_params, _ = cassowaries_optimization(build_and_evaluate_model, np.array(lb), np.array(ub))

# Build the final model with optimal parameters
units1_optimal = int(optimal_params[0])
units2_optimal = int(optimal_params[1])
batch_size_optimal = int(optimal_params[2])

model = Sequential()
model.add(GRU(units=units1_optimal, return_sequences=True, input_shape=(train_features.shape[1], train_features.shape[2])))
model.add(GRU(units=units2_optimal))
model.add(Dense(1))
model.compile(optimizer='adam', loss='mean_squared_error')

early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
history = model.fit(
    train_features,
    train_target,
    epochs=50,
    batch_size=batch_size_optimal,
    validation_data=(test_features, test_target),
    callbacks=[early_stopping],
    verbose=1
)

# Make predictions
predictions = model.predict(test_features)
actuals = scaler.inverse_transform(test_target)
predictions = scaler.inverse_transform(predictions)

# Calculate metrics
mae = mean_absolute_error(actuals, predictions)
rmse = np.sqrt(mean_squared_error(actuals, predictions))
r2 = r2_score(actuals, predictions)

# Print metrics
print(f"MAE: {mae}")
print(f"RMSE: {rmse}")
print(f"R^2: {r2}")

# Plot the training and validation loss
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()

# Plot actual vs predicted values
plt.figure(figsize=(10, 6))
plt.plot(actuals, label='Actual Prices')
plt.plot(predictions, label='Predicted Prices')
plt.title('Actual vs Predicted Prices')
plt.xlabel('Time Steps')
plt.ylabel('Price')
plt.legend()
plt.show()
```
